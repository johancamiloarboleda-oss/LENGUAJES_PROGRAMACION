## Máximos y Mínimos Históricos

Vamos a calcular el precio de cierre máximo y mínimo histórico.

```python
precio_maximo = df['close'].max()
precio_minimo = df['close'].min()

print(f"Precio de Cierre Máximo Histórico: {precio_maximo:.2f}")
print(f"Precio de Cierre Mínimo Histórico: {precio_minimo:.2f}")
```

## Gráfica de Máximos y Mínimos Históricos

Visualicemos la serie de precios de cierre y marquemos los puntos de máximo y mínimo histórico.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.plot(df['date'], df['close'], label='Precio de Cierre')
plt.scatter(df.loc[df['close'].idxmax()]['date'], precio_maximo, color='green', s=100, zorder=5, label=f'Máximo Histórico: {precio_maximo:.2f}')
plt.scatter(df.loc[df['close'].idxmin()]['date'], precio_minimo, color='red', s=100, zorder=5, label=f'Mínimo Histórico: {precio_minimo:.2f}')
plt.title('Serie de Precios de Cierre con Máximos y Mínimos Históricos')
plt.xlabel('Fecha')
plt.ylabel('Precio de Cierre')
plt.legend()
plt.grid(True)
plt.show()
```

## Rendimientos Diarios

Visualicemos la serie de rendimientos diarios para observar su fluctuación.

```python
plt.figure(figsize=(12, 6))
plt.plot(df['date'][df['rendimiento_diario'] > 0], df['rendimiento_diario'][df['rendimiento_diario'] > 0], 'o', label='Rendimiento Positivo', color='green', markersize=3)
plt.plot(df['date'][df['rendimiento_diario'] < 0], df['rendimiento_diario'][df['rendimiento_diario'] < 0], 'o', label='Rendimiento Negativo', color='red', markersize=3)
plt.plot(df['date'][df['rendimiento_diario'] == 0], df['rendimiento_diario'][df['rendimiento_diario'] == 0], 'o', label='Rendimiento Neutro', color='gray', markersize=3)
plt.title('Serie de Rendimientos Diarios Coloreados')
plt.xlabel('Fecha')
plt.ylabel('Rendimiento Diario')
plt.legend()
plt.grid(True)
plt.axhline(0, color='black', linestyle='--', linewidth=0.8) # Línea de referencia en cero
plt.show()
```

1.  **¿El activo es volátil? ¿Por qué?**

    Observando el gráfico de la distribución de rendimientos diarios y el valor de la volatilidad diaria (0.0146), podemos decir que el activo muestra una volatilidad moderada. Activos con mayores rendimientos extremos (positivos o negativos) y una mayor desviación estándar se consideran más volátiles.

2.  **¿Tiene tendencia clara?**

    Para determinar una tendencia clara, hay que analizar el gráfico de la Serie de Precios de Cierre. Una tendencia alcista se vería como un movimiento sostenido hacia arriba a lo largo del tiempo, mientras que una tendencia bajista sería un movimiento sostenido hacia abajo. Si el precio se mueve lateralmente con fluctuaciones, podríamos hablar de una tendencia lateral o consolidación.

3. **¿Hay eventos con movimientos fuertes?**

    Los gráficos de Serie de Precios y Rendimientos Diarios son clave aquí. 4.  **¿Es buena inversión a corto/largo plazo?**

    Como inteligencia artificial, no puedo ofrecer asesoramiento de inversión. Sin embargo, el análisis que hemos realizado proporciona datos útiles:

    A largo plazo: Una tendencia alcista clara en el gráfico de precios es deseable, combinada con un rendimiento promedio positivo. La volatilidad también es un factor, ya que una alta volatilidad a largo plazo podría implicar mayores oscilaciones.
